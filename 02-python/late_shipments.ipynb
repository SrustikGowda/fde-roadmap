## 1. Get Meera's file

This cell downloads `shipments.csv` (2,000 made-up shipments) so the program below can open it.

```python
import urllib.request

urllib.request.urlretrieve(
    "https://raw.githubusercontent.com/SrustikGowda/fde-roadmap/main/02-python/shipments.csv",
    "shipments.csv",
)
print("Meera's file is ready")
```

## 2. The program from the video

It reads the file one row at a time, counts the late shipments, and counts the ones not delivered yet.

```python
import csv

late = 0
waiting = 0

file = open("shipments.csv")
for row in csv.DictReader(file):
    if row["days_taken"] == "":
        waiting = waiting + 1
    elif int(row["days_taken"]) > int(row["days_promised"]):
        late = late + 1

print("Late:", late)
print("Not delivered yet:", waiting)
```
